```python
import pandas as pd
import nltk
from nltk.corpus import stopwords

# Reload the dataset (as it might have been reset in the previous error)
df = pd.read_csv("C:\\Users\\Chetti Rishika\\Downloads\\social_media.csv")

# Reapply previous cleaning steps to df
likes_median = df['likes'].median()
shares_median = df['shares'].median()
df['likes'] = df['likes'].fillna(likes_median)
df['shares'] = df['shares'].fillna(shares_median)
df['post_text'] = df['post_text'].str.lower()
df['post_text'] = df['post_text'].str.replace(r'http\S+|www\S+|https\S+', '', regex=True)
df['post_text'] = df['post_text'].str.replace(r'@\w+', '', regex=True)
df['post_text'] = df['post_text'].str.replace(r'#\w+', '', regex=True)
df['post_text'] = df['post_text'].str.replace(r'[^\w\s]', '', regex=True)
df['post_text'] = df['post_text'].str.replace(r'\d+', '', regex=True)
df['post_text'] = df['post_text'].str.strip()
df['post_text'] = df['post_text'].str.replace(r'\s+', ' ', regex=True)

# Download stopwords if not already downloaded
nltk.download('stopwords')

# Create a set of English stopwords
stop_words = set(stopwords.words('english'))

# Define a function to remove stopwords
def remove_stopwords(text):
    words = text.split()
    filtered_words = [word for word in words if word not in stop_words]
    return " ".join(filtered_words)

# Apply the function to the 'post_text' column
df['post_text'] = df['post_text'].apply(remove_stopwords)

print("Common English stopwords removed from 'post_text'.")
print(df.head())
```

```python
df['timestamp'] = pd.to_datetime(df['timestamp'])
df['hour'] = df['timestamp'].dt.hour
df['weekday'] = df['timestamp'].dt.day_name()

print("Extracted 'hour' and 'weekday' from 'timestamp'.")
print(df.head())
```

```python
print("Number of duplicate rows before removal:")
print(df.duplicated().sum())
```

```python
print("Number of duplicate posts based on 'post_text' before removal:")
print(df.duplicated(subset=['post_text']).sum())

df.drop_duplicates(subset=['post_text'], inplace=True)

print("DataFrame after removing duplicate posts based on 'post_text':")
print(df.head())
print(f"Shape of the DataFrame after removing duplicate posts: {df.shape}")
```

```python
spam_keywords = ['free', 'win', 'promo', 'discount']

# Identify spam posts
# The str.contains() method allows checking for multiple patterns using regex OR (|)
spam_pattern = '|'.join(spam_keywords)
is_spam = df['post_text'].str.contains(spam_pattern, case=False, na=False)

print(f"Number of spam posts identified: {is_spam.sum()}")

# Remove spam posts
df = df[~is_spam]

print("DataFrame after removing spam posts:")
print(df.head())
print(f"Shape of the DataFrame after removing duplicate and spam posts: {df.shape}")
```
