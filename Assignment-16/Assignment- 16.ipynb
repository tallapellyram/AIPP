```python
import sqlite3
import pandas as pd

# Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Create the employees table
cursor.execute('''
    CREATE TABLE employees (
        id INTEGER PRIMARY KEY,
        name TEXT NOT NULL,
        department TEXT NOT NULL,
        salary REAL NOT NULL
    )
''')

# Insert sample data
employees_data = [
    (1, 'Alice Smith', 'Sales', 75000),
    (2, 'Bob Johnson', 'Marketing', 60000),
    (3, 'Charlie Brown', 'Sales', 80000),
    (4, 'David Lee', 'IT', 90000),
    (5, 'Eve Davis', 'Marketing', 65000),
    (6, 'Frank White', 'IT', 95000),
    (7, 'Grace Taylor', 'Sales', 82000),
    (8, 'Henry Wilson', 'HR', 55000),
    (9, 'Amy Green', 'Marketing', 70000),
    (10, 'Andrew Clark', 'IT', 88000)
]

cursor.executemany('INSERT INTO employees (id, name, department, salary) VALUES (?, ?, ?, ?)', employees_data)
conn.commit()

print("Database and 'employees' table created with sample data.")

```

```python
query_starts_with_a = "SELECT name, department, salary FROM employees WHERE name LIKE 'A%'"
employees_starts_with_a = pd.read_sql_query(query_starts_with_a, conn)
print("Employees whose name starts with 'A':")
display(employees_starts_with_a)
```

```python
print("Original employee departments:")
display(pd.read_sql_query("SELECT id, name, department FROM employees", conn))

# Update department names
cursor.execute("UPDATE employees SET department = 'Customer Relations' WHERE department = 'Sales'")
cursor.execute("UPDATE employees SET department = 'Digital Outreach' WHERE department = 'Marketing'")
conn.commit()

print("\nUpdated employee departments:")
display(pd.read_sql_query("SELECT id, name, department FROM employees", conn))
```

```python
query_top_3 = "SELECT name, department, salary FROM employees ORDER BY salary DESC LIMIT 3"
top_3_employees = pd.read_sql_query(query_top_3, conn)
print("Top 3 highest-paid employees:")
display(top_3_employees)
```

```python
cursor.execute('''
    CREATE TABLE projects (
        project_id INTEGER PRIMARY KEY,
        project_name TEXT NOT NULL,
        employee_id INTEGER,
        FOREIGN KEY (employee_id) REFERENCES employees(id)
    )
''')

# Insert sample project data
projects_data = [
    (101, 'Website Redesign', 1),
    (102, 'New Marketing Campaign', 2),
    (103, 'Internal Tools Development', 4),
    (104, 'Sales Strategy 2024', 7),
    (105, 'Cloud Migration', 6),
    (106, 'HR System Upgrade', 8),
    (107, 'Customer Feedback Analysis', 5)
]
cursor.executemany('INSERT INTO projects (project_id, project_name, employee_id) VALUES (?, ?, ?)', projects_data)
conn.commit()

print("Created 'projects' table and linked it with 'employees'.")

display(pd.read_sql_query("SELECT * FROM projects", conn))
```

```python
cursor.execute("ALTER TABLE employees ADD COLUMN email TEXT")
conn.commit()

# Populate email addresses for existing employees (example)
email_updates = [
    ('alice.smith@example.com', 1),
    ('bob.johnson@example.com', 2),
    ('charlie.brown@example.com', 3),
    ('david.lee@example.com', 4),
    ('eve.davis@example.com', 5),
    ('frank.white@example.com', 6),
    ('grace.taylor@example.com', 7),
    ('henry.wilson@example.com', 8),
    ('amy.green@example.com', 9),
    ('andrew.clark@example.com', 10)
]
cursor.executemany("UPDATE employees SET email = ? WHERE id = ?", email_updates)
conn.commit()

print("Added 'email' column and populated it for existing employees.")
display(pd.read_sql_query("SELECT id, name, department, email, salary FROM employees", conn))
```
