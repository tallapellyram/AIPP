```python
import requests
import json

def get_weather(city_name):
    api_key = "YOUR_API_KEY"  # Replace with your actual OpenWeatherMap API key
    base_url = "http://api.openweathermap.org/data/2.5/weather"
    
    try:
        # Build request URL
        complete_url = f"{base_url}?q={city_name}&appid={api_key}&units=metric"
        
        # Send GET request with timeout
        response = requests.get(complete_url, timeout=5)
        
        # Raise HTTPError if status code is not 200
        response.raise_for_status()
        
        # Parse JSON response
        data = response.json()
        
        # Print formatted JSON output
        print("Weather Details (Pretty JSON):")
        print(json.dumps(data, indent=4))
    
    except requests.exceptions.HTTPError as http_err:
        print("Error: Could not connect to API. Check your API key or city name.")
        print(f"Details: {http_err}")
    except requests.exceptions.ConnectionError:
        print("Error: Could not connect to API. Check your network connection.")
    except requests.exceptions.Timeout:
        print("Error: Request timed out. Please try again later.")
    except Exception as err:
        print("Error: An unexpected error occurred.")
        print(f"Details: {err}")

# Example usage:
get_weather("London")
```
