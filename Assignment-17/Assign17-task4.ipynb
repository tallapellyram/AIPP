```python
import pandas as pd
import os

file_path = "C:\\Users\\Chetti Rishika\\Downloads\\movie_reviews-1.csv"

# Load the dataset into a pandas DataFrame
try:
    df = pd.read_csv(file_path)

    # Display the first 5 rows of the DataFrame
    print("First 5 rows of the DataFrame:")
    print(df.head())

    # Print the shape of the DataFrame
    print("\nShape of the DataFrame:")
    print(df.shape)

    # Display the column names and their data types
    print("\nColumn names and their data types:")
    print(df.info())

except FileNotFoundError:
    print(f"### File Upload Required\n\nThe '{os.path.basename(file_path)}' file is not present in the current Colab environment. "
          "To proceed with the task, please upload the '{os.path.basename(file_path)}' file to the Colab session. "
          "You can do this by clicking the 'Files' icon on the left sidebar, then clicking 'Upload' and selecting your file. "
          "Once the file has been uploaded, please re-run this code cell.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
df['review_text'] = df['review_text'].str.lower()
df['review_text'] = df['review_text'].str.replace(r'<[^>]*>', '', regex=True)

print("Cleaned 'review_text' column:")
print(df.head())
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

print("TfidfVectorizer imported.")
```

```python
tfidf_vectorizer = TfidfVectorizer()
tfidf_matrix = tfidf_vectorizer.fit_transform(df['review_text'])

tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=tfidf_vectorizer.get_feature_names_out())
df = pd.concat([df, tfidf_df], axis=1)

print("DataFrame with TF-IDF features:")
print(df.head())
```

```python
rating_median = df['rating'].median()
df['rating'] = df['rating'].fillna(rating_median)

print("Missing values in 'rating' after median imputation:")
print(df['rating'].isnull().sum())

print("DataFrame after handling missing 'rating' values:")
print(df.head())
```

```python
min_rating = df['rating'].min()
max_rating = df['rating'].max()
df['rating_normalized'] = (df['rating'] - min_rating) / (max_rating - min_rating)

print("DataFrame with normalized 'rating' column (0-1 scale):")
print(df.head())
```
