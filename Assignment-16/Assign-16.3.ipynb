```python
import sqlite3

# 1. Connect to an in-memory SQLite database
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# 2. Execute the DDL statement for the 'Books' table
cursor.execute('''
    CREATE TABLE Books (
        BookID INTEGER PRIMARY KEY,
        Title TEXT NOT NULL,
        Author TEXT NOT NULL,
        ISBN TEXT NOT NULL UNIQUE,
        PublicationYear INTEGER,
        Status TEXT NOT NULL
    )
''')
# 3. Commit the transaction for 'Books'
conn.commit()

# 4. Execute the DDL statement for the 'Members' table
cursor.execute('''
    CREATE TABLE Members (
        MemberID INTEGER PRIMARY KEY NOT NULL UNIQUE,
        Name TEXT NOT NULL,
        Address TEXT,
        PhoneNumber TEXT UNIQUE,
        Email TEXT UNIQUE
    )
''')
# 5. Commit the transaction for 'Members'
conn.commit()

# 6. Execute the DDL statement for the 'Loans' table
cursor.execute('''
    CREATE TABLE Loans (
        LoanID INTEGER PRIMARY KEY,
        BookID INTEGER NOT NULL,
        MemberID INTEGER NOT NULL,
        LoanDate TEXT NOT NULL,
        ReturnDate TEXT,
        FOREIGN KEY (BookID) REFERENCES Books(BookID),
        FOREIGN KEY (MemberID) REFERENCES Members(MemberID)
    )
''')
# 7. Commit the transaction for 'Loans'
conn.commit()

# 8. Print a message confirming that the database and tables have been created successfully.
print("Database and all tables ('Books', 'Members', 'Loans') created successfully.")
```

```python
print("\n--- Books Table Schema ---")
cursor.execute("PRAGMA table_info(Books)")
books_schema = cursor.fetchall()
for col in books_schema:
    print(f"Column: {col[1]}, Type: {col[2]}, NotNull: {bool(col[3])}, PrimaryKey: {bool(col[5])}")

print("\n--- Members Table Schema ---")
cursor.execute("PRAGMA table_info(Members)")
members_schema = cursor.fetchall()
for col in members_schema:
    print(f"Column: {col[1]}, Type: {col[2]}, NotNull: {bool(col[3])}, PrimaryKey: {bool(col[5])}")

print("\n--- Loans Table Schema ---")
cursor.execute("PRAGMA table_info(Loans)")
loans_schema = cursor.fetchall()
for col in loans_schema:
    print(f"Column: {col[1]}, Type: {col[2]}, NotNull: {bool(col[3])}, PrimaryKey: {bool(col[5])}")
```
