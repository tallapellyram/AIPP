```python
import requests
import json
import os

def get_weather(city_name):
    api_key = "YOUR_API_KEY"  # Replace with your actual OpenWeatherMap API key
    base_url = "http://api.openweathermap.org/data/2.5/weather"
    
    try:
        # Build request URL dynamically based on city name
        complete_url = f"{base_url}?q={city_name}&appid={api_key}&units=metric"
        
        # Send GET request
        response = requests.get(complete_url, timeout=5)
        
        # Handle invalid city (404 response)
        if response.status_code == 404:
            print("Error: City not found. Please enter a valid city.")
            return
        
        # Raise HTTPError for other bad responses
        response.raise_for_status()
        
        # Parse JSON response
        data = response.json()
        
        # Extract specific fields
        city = data["name"]
        temperature = data["main"]["temp"]
        humidity = data["main"]["humidity"]
        description = data["weather"][0]["description"].capitalize()
        
        # Display in user-friendly format
        print(f"City: {city}")
        print(f"Temperature: {temperature}°C")
        print(f"Humidity: {humidity}%")
        print(f"Weather: {description}")
        
        # Prepare result dictionary
        result = {
            "city": city,
            "temp": temperature,
            "humidity": humidity,
            "weather": description
        }
        
        # File path
        file_name = "results.json"
        
        # If file exists, load existing data
        if os.path.exists(file_name):
            with open(file_name, "r") as file:
                try:
                    results = json.load(file)
                except json.JSONDecodeError:
                    results = []
        else:
            results = []
        
        # Append new result
        results.append(result)
        
        # Save back to file
        with open(file_name, "w") as file:
            json.dump(results, file, indent=4)
    
    except requests.exceptions.HTTPError as http_err:
        print("Error: Could not connect to API. Check your API key or city name.")
        print(f"Details: {http_err}")
    except requests.exceptions.ConnectionError:
        print("Error: Could not connect to API. Check your network connection.")
    except requests.exceptions.Timeout:
        print("Error: Request timed out. Please try again later.")
    except Exception as err:
        print("Error: An unexpected error occurred.")
        print(f"Details: {err}")

# Example usage:
get_weather("London")
get_weather("New York")
get_weather("xyz123")  # Invalid city
```
