```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("C:\\Users\\Chetti Rishika\\Downloads\\financial_data.csv", parse_dates=["date"])

# 1. Handle missing values
# Closing price: forward fill, then backward fill
df["closing_price"] = df["closing_price"].fillna(method="ffill")
df["closing_price"] = df["closing_price"].fillna(method="bfill")

# Volume: fill missing with median
median_volume = df["volume"].median()
df["volume"] = df["volume"].fillna(median_volume)

# 2. Create lag features
df["return_1d"] = df["closing_price"].pct_change(1)   # 1-day return
df["return_7d"] = df["closing_price"].pct_change(7)   # 7-day return

# 3. Normalize volume using log-scaling
df["log_volume"] = np.log(df["volume"] + 1)

# 4. Detect outliers in closing_price using IQR
Q1 = df["closing_price"].quantile(0.25)
Q3 = df["closing_price"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df["closing_price_outlier"] = (
    (df["closing_price"] < lower_bound) | (df["closing_price"] > upper_bound)
)

# Final dataset ready for forecasting
print(df.head(15))

# Optionally save to new CSV
df.to_csv("financial_data_preprocessed.csv", index=False)
```
